```python
import pandas as pd

marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
)

print(marks)
```

```python
import pandas as pd

marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
)

print("1st rank mark:")
print(marks[0])

print("\nTop 3 ranks using loc:")
print(marks.loc[['Rank1', 'Rank2', 'Rank3']])

print("\n3rd rank mark using iloc:")
print(marks.iloc[2])

print("\nRanks with marks greater than 90:")
print(marks[marks > 90])
```

```python
import pandas as pd

marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
)

marks.loc['Rank1'] = 100
print("Updated marks:")
print(marks)


marks = marks.drop('Rank5')
print("\nMarks after removing the last rank:")
print(marks)


cgpa = marks / 10
print("\nCGPA:")
print(cgpa)
```

```python
import pandas as pd

data = {
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'ProductCategory': [
        'Electronics', 'Clothing', 'Electronics', 'Furniture', 'Clothing',
        'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'
    ],
    'Region': [
        'North', 'South', 'North', 'East', 'West',
        'North', 'East', 'West', 'South', 'North'
    ],
    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
}

transactions = pd.DataFrame(data)

print(transactions)
```

```python
print("Full DataFrame:")
print(transactions)

print("\nFirst 5 rows:")
print(transactions.head())

print("\nLast 5 rows:")
print(transactions.tail())

print("\nShape (rows, columns):")
print(transactions.shape)

print("\nColumn names:")
print(transactions.columns)

print("\nData types:")
print(transactions.dtypes)

print("\nBasic information:")
transactions.info()


print("\nProductCategory and Amount:")
print(transactions[['ProductCategory', 'Amount']])

print("\nLast 3 columns:")
print(transactions.iloc[:, -3:])

print("\nNorth region transactions with Amount > 200:")
print(transactions[
    (transactions['Region'] == 'North') &
    (transactions['Amount'] > 200)
])


print("\nProduct category counts:")
print(transactions['ProductCategory'].value_counts())


print("\nUnique regions:")
print(transactions['Region'].unique())


print("\nMean Amount for each region:")
print(transactions.groupby('Region')['Amount'].mean())
```

```python
transactions.loc[transactions['TransactionID'] == 102, 'Amount'] = 165

print("1. After modifying Amount:")
print(transactions)

transactions['Discount'] = transactions['Amount'] * 0.10

print("\n2. After adding Discount:")
print(transactions)

transactions = transactions[transactions['TransactionID'] != 109]

print("\n3. After removing TransactionID 109:")
print(transactions)

transactions.drop(columns=['Discount'], inplace=True)

print("\n4. Final DataFrame:")
print(transactions)
```
