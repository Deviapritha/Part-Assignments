```python
age_list = [24, 25, 26, 27, 28]
name_list = ["Anu", "Ravi", "Priya", "Arun", "Meena"]
print(age_list)
print(name_list)
```

```python
name_list.append("Yazhini")
print("After append:", name_list)
```

```python
age_list.insert(2, 30)
print("After insert:", age_list)
```

```python
name_list.remove("Yazhini")
print("remove:", name_list)
```

```python
age_list.pop()
print("After pop", age_list)
```

```python
age_list.extend([29, 30, 26])
print("after extend", age_list)
```

```python
age_list.sort(reverse=True)
print("Descending order:", age_list)
```

```python
max_age = max(age_list)
min_age = min(age_list)
total_age = sum(age_list)

print("Maximum age:", max_age)
print("Minimum age:", min_age)
print("Sum of ages:", total_age)
```

```python
## 3. Accessing List Elements:
print("First element:", name_list[0])
print("Last element:", name_list[-1])
print("Elements from index 2 to 4:", name_list[2:5])
print("Reverse order:", name_list[::-1])
```

```python
## Dictionary (Creation, Modification and Access):
student_marks = {
    "Anu": 75,
    "Ravi": 88,
    "Priya": 92,
    "Arun": 68,
    "Meena": 81
}
print("Priya's mark:", student_marks["Priya"])
```

```python
student_marks["Janani"] = 80
print("After adding Janani:", student_marks)
```

```python
student_marks["Anu"] = 82
print("After updating Anu's mark:", student_marks)
print("Keys:", student_marks.keys())
print("Values:", student_marks.values())
print("Items:", student_marks.items())
```

```python
## Sets (Operations):
my_set = {'a', 'e', 'i', 'o', 'u', 'a', 'a', 'i'}
print(my_set)

## A set does not allow duplicate values. Therefore, the repeated 'a' and 'i' are automatically removed.
## Also, sets are unordered, so the output order may be different when you run the code.
```

```python
my_set[4] = 's'

## Sets do not support indexing because they are unordered collections. Therefore, you cannot access or modify an element using an index such as my_set[4].
```

```python
set1 = {1, 3, 5, 7, 9}

set2 = {2, 3, 5, 8, 10}

print("Set 1:", set1)
print("Set 2:", set2)
```

```python
union_set = set1.union(set2)
intersection_set = set1.intersection(set2)
print("Union:", union_set)
print("Intersection:", intersection_set)
```
